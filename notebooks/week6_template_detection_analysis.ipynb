```python
# Contour-Based Grid Detection Implementation
def contour_grid_detection(image, config):
    """
    Detect OMR grid menggunakan contour analysis
    
    Args:
        image: Input OMR image (grayscale)
        config: Detection configuration
    
    Returns:
        dict: Detection results dengan grid coordinates dan confidence
    """
    import time
    start_time = time.time()
    
    # Preprocessing untuk contour detection
    # 1. Gaussian blur untuk noise reduction
    blurred = cv2.GaussianBlur(image, (5, 5), 0)
    
    # 2. Adaptive thresholding untuk robust binarization
    binary = cv2.adaptiveThreshold(
        blurred, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, 
        cv2.THRESH_BINARY_INV, 11, 2
    )
    
    # 3. Morphological operations untuk cleanup
    kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (3, 3))
    cleaned = cv2.morphologyEx(binary, cv2.MORPH_CLOSE, kernel)
    cleaned = cv2.morphologyEx(cleaned, cv2.MORPH_OPEN, kernel)
    
    # Find contours
    contours, hierarchy = cv2.findContours(cleaned, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    
    # Filter contours berdasarkan area dan aspect ratio
    min_area = config.detection.get('min_area', 5000)
    max_area = config.detection.get('max_area', 200000)
    aspect_ratio_range = config.detection.get('aspect_ratio_range', (0.3, 3.0))
    
    candidate_contours = []
    
    for contour in contours:
        area = cv2.contourArea(contour)
        
        # Filter by area
        if area < min_area or area > max_area:
            continue
            
        # Calculate bounding rectangle
        x, y, w, h = cv2.boundingRect(contour)
        aspect_ratio = w / h if h > 0 else 0
        
        # Filter by aspect ratio
        if aspect_ratio < aspect_ratio_range[0] or aspect_ratio > aspect_ratio_range[1]:
            continue
            
        # Calculate rectangularity (how close to rectangle)
        rect_area = w * h
        rectangularity = area / rect_area if rect_area > 0 else 0
        
        if rectangularity > 0.5:  # Threshold untuk rectangularity
            candidate_contours.append({
                'contour': contour,
                'area': area,
                'bbox': (x, y, w, h),
                'aspect_ratio': aspect_ratio,
                'rectangularity': rectangularity,
                'score': rectangularity * (area / max_area)
            })
    
    # Select best candidate
    if not candidate_contours:
        return {\n            'success': False,\n            'grid_coordinates': None,\n            'confidence': 0.0,\n            'processing_time': time.time() - start_time,\n            'method': 'contour',\n            'candidates_found': 0\n        }\n    \n    # Sort by score dan select best\n    candidate_contours.sort(key=lambda x: x['score'], reverse=True)\n    best_candidate = candidate_contours[0]\n    \n    # Extract grid coordinates\n    x, y, w, h = best_candidate['bbox']\n    grid_coordinates = [x, y, x + w, y + h]  # [x1, y1, x2, y2]\n    \n    # Calculate confidence score\n    confidence = min(1.0, best_candidate['score'] * 2)  # Normalize to 0-1\n    \n    processing_time = time.time() - start_time\n    \n    return {\n        'success': True,\n        'grid_coordinates': grid_coordinates,\n        'confidence': confidence,\n        'processing_time': processing_time,\n        'method': 'contour',\n        'candidates_found': len(candidate_contours),\n        'best_candidate': best_candidate,\n        'preprocessing_steps': ['gaussian_blur', 'adaptive_threshold', 'morphological_ops']\n    }\n\n# Test contour detection pada semua test images\nprint(\"🔍 Testing Contour-Based Grid Detection:\")\ncontour_results = {}\n\nfor name, img in test_images.items():\n    print(f\"\\n  📝 Processing {name} image...\")\n    result = contour_grid_detection(img, td_config)\n    contour_results[name] = result\n    \n    status = \"SUCCESS\" if result['success'] else \"FAILED\"\n    conf_str = f\"conf: {result['confidence']:.3f}\" if result['success'] else \"conf: 0.000\"\n    time_str = f\"time: {result['processing_time']:.3f}s\"\n    candidates_str = f\"candidates: {result['candidates_found']}\"\n    \n    print(f\"    Status: {status}, {conf_str}, {time_str}, {candidates_str}\")\n\n# Visualize contour detection results\nfig, axes = plt.subplots(2, 4, figsize=(20, 10))\n\nfor idx, (name, img) in enumerate(test_images.items()):\n    # Original image\n    axes[0, idx].imshow(img, cmap='gray')\n    axes[0, idx].set_title(f'{name.title()} Original')\n    axes[0, idx].axis('off')\n    \n    # Detection result\n    result_img = img.copy()\n    result = contour_results[name]\n    \n    if result['success']:\n        # Draw detected grid\n        coords = result['grid_coordinates']\n        cv2.rectangle(result_img, (coords[0], coords[1]), (coords[2], coords[3]), 128, 3)\n        \n        # Add confidence text\n        cv2.putText(result_img, f\"Conf: {result['confidence']:.3f}\", \n                   (coords[0], coords[1]-10), cv2.FONT_HERSHEY_SIMPLEX, 0.7, 128, 2)\n    \n    axes[1, idx].imshow(result_img, cmap='gray')\n    axes[1, idx].set_title(f'Contour Detection\\n(Success: {result[\"success\"]})')\n    axes[1, idx].axis('off')\n\nplt.suptitle('Contour-Based Grid Detection Results', fontsize=16)\nplt.tight_layout()\nplt.show()\n\n# Performance summary\nprint(f\"\\n📊 Contour Detection Performance Summary:\")\nsuccess_count = sum(1 for r in contour_results.values() if r['success'])\ntotal_count = len(contour_results)\navg_confidence = np.mean([r['confidence'] for r in contour_results.values()])\navg_time = np.mean([r['processing_time'] for r in contour_results.values()])\n\nprint(f\"  Success Rate: {success_count}/{total_count} ({success_count/total_count*100:.1f}%)\")\nprint(f\"  Average Confidence: {avg_confidence:.3f}\")\nprint(f\"  Average Processing Time: {avg_time:.3f}s\")\nprint(f\"  Average Candidates Found: {np.mean([r['candidates_found'] for r in contour_results.values()]):.1f}\")"
```

```python
# Create synthetic OMR test images untuk demonstration
def create_synthetic_omr_image(width=800, height=600, noise_level=0):
    """
    Create synthetic OMR answer sheet untuk testing
    
    Args:
        width, height: Image dimensions
        noise_level: 0=perfect, 1=slight noise, 2=heavy noise
    
    Returns:
        numpy array: Synthetic OMR image
    """
    # Create white background
    image = np.ones((height, width), dtype=np.uint8) * 255
    
    # OMR Grid parameters (3 columns x 20 rows = 60 questions)
    cols, rows = 3, 20
    cell_width = 180
    cell_height = 25
    
    # Grid starting position
    start_x, start_y = 50, 50
    
    # Draw main grid lines
    # Horizontal lines
    for i in range(rows + 1):
        y = start_y + i * cell_height
        cv2.line(image, (start_x, y), (start_x + cols * cell_width, y), 0, 2)
    
    # Vertical lines  
    for j in range(cols + 1):
        x = start_x + j * cell_width
        cv2.line(image, (x, start_y), (x, start_y + rows * cell_height), 0, 2)
    
    # Add bubbles in each cell (A, B, C, D, E options)
    bubble_radius = 7
    bubbles_per_cell = 5
    
    for row in range(rows):
        for col in range(cols):
            cell_x = start_x + col * cell_width
            cell_y = start_y + row * cell_height
            
            # Draw 5 option bubbles
            for bubble_idx in range(bubbles_per_cell):
                bubble_x = cell_x + 15 + bubble_idx * 30
                bubble_y = cell_y + cell_height // 2
                
                # Draw bubble outline
                cv2.circle(image, (bubble_x, bubble_y), bubble_radius, 0, 2)
                
                # Randomly fill some bubbles (20% chance)
                if np.random.random() < 0.2:
                    cv2.circle(image, (bubble_x, bubble_y), bubble_radius-2, 0, -1)
    
    # Add noise berdasarkan level
    if noise_level > 0:
        # Gaussian noise
        noise = np.random.normal(0, 5 * noise_level, image.shape)
        image = np.clip(image.astype(float) + noise, 0, 255).astype(np.uint8)
        
        # Random spots
        for _ in range(20 * noise_level):
            x = np.random.randint(0, width)
            y = np.random.randint(0, height)
            cv2.circle(image, (x, y), np.random.randint(1, 3), 0, -1)
    
    return image

# Create test images dengan berbagai conditions
test_images = {}

print("🎨 Creating synthetic test images...")

# Perfect OMR image
test_images['perfect'] = create_synthetic_omr_image(noise_level=0)
print("  ✅ Perfect OMR image created")

# Slightly noisy image
test_images['noisy'] = create_synthetic_omr_image(noise_level=1)
print("  ✅ Noisy OMR image created")

# Heavily distorted image  
test_images['distorted'] = create_synthetic_omr_image(noise_level=2)
print("  ✅ Distorted OMR image created")

# Rotated image
perfect_img = test_images['perfect']
height, width = perfect_img.shape
center = (width // 2, height // 2)
rotation_matrix = cv2.getRotationMatrix2D(center, 5, 1.0)  # 5 degree rotation
test_images['rotated'] = cv2.warpAffine(perfect_img, rotation_matrix, (width, height), borderValue=255)
print("  ✅ Rotated OMR image created")

print(f"\n📊 Test Images Summary:")
for name, img in test_images.items():
    print(f"  {name}: {img.shape} - {img.dtype}")

# Visualize test images
fig, axes = plt.subplots(2, 2, figsize=(15, 10))
axes = axes.ravel()

for idx, (name, img) in enumerate(test_images.items()):
    axes[idx].imshow(img, cmap='gray')
    axes[idx].set_title(f'{name.title()} OMR Image')
    axes[idx].axis('off')

plt.suptitle('Synthetic OMR Test Images untuk Template Detection', fontsize=16)
plt.tight_layout()
plt.show()

# Save test images untuk future reference
for name, img in test_images.items():
    cv2.imwrite(str(WEEK6_OUTPUT / f'synthetic_omr_{name}.jpg'), img)

print(f"✅ Test images saved to {WEEK6_OUTPUT}")
```

## Setup Development Configuration

```python
# Template Detection Configuration
@dataclass
class TemplateDetectionConfig:
    """Configuration class untuk Template Detection parameters"""
    
    # Contour Detection Parameters
    contour_min_area: int = 5000
    contour_max_area: int = 200000
    contour_aspect_ratio_min: float = 0.3
    contour_aspect_ratio_max: float = 3.0
    contour_rectangularity_threshold: float = 0.7
    
    # Hough Transform Parameters
    hough_rho_resolution: float = 1.0
    hough_theta_resolution: float = np.pi/180
    hough_threshold: int = 100
    hough_min_line_length: int = 50
    hough_max_line_gap: int = 10
    
    # Template Matching Parameters
    template_scale_range: Tuple[float, float] = (0.7, 1.3)
    template_scale_step: float = 0.1
    template_rotation_range: Tuple[int, int] = (-20, 20)
    template_rotation_step: int = 5
    template_match_threshold: float = 0.7
    
    # Detection Fusion Parameters
    fusion_confidence_threshold: float = 0.6
    fusion_geometric_tolerance: float = 0.1
    fusion_consensus_weight: Dict[str, float] = None
    
    # Performance Parameters
    processing_timeout: int = 30  # seconds
    max_detection_attempts: int = 3
    
    def __post_init__(self):
        if self.fusion_consensus_weight is None:
            self.fusion_consensus_weight = {
                'contour': 0.4,
                'hough': 0.3,
                'template': 0.3
            }

# Create default configuration
config = TemplateDetectionConfig()

# Save configuration for reproducibility
config_dict = {
    'contour_parameters': {
        'min_area': config.contour_min_area,
        'max_area': config.contour_max_area,
        'aspect_ratio_range': [config.contour_aspect_ratio_min, config.contour_aspect_ratio_max],
        'rectangularity_threshold': config.contour_rectangularity_threshold
    },
    'hough_parameters': {
        'rho_resolution': config.hough_rho_resolution,
        'theta_resolution': float(config.hough_theta_resolution),
        'threshold': config.hough_threshold,
        'min_line_length': config.hough_min_line_length,
        'max_line_gap': config.hough_max_line_gap
    },
    'template_parameters': {
        'scale_range': config.template_scale_range,
        'scale_step': config.template_scale_step,
        'rotation_range': config.template_rotation_range,
        'rotation_step': config.template_rotation_step,
        'match_threshold': config.template_match_threshold
    },
    'fusion_parameters': {
        'confidence_threshold': config.fusion_confidence_threshold,
        'geometric_tolerance': config.fusion_geometric_tolerance,
        'consensus_weights': config.fusion_consensus_weight
    }
}

with open(WEEK6_OUTPUT / 'template_detection_config.json', 'w') as f:
    json.dump(config_dict, f, indent=2)

print("✅ Template Detection Configuration initialized")
print(f"Configuration saved to: {WEEK6_OUTPUT / 'template_detection_config.json'}")
```

## Development Utilities Setup

```python
# Visualization utilities untuk detection debugging
def setup_visualization_style():
    """Setup matplotlib style untuk consistent visualization"""
    plt.style.use('default')
    sns.set_palette("husl")
    
    # Custom color scheme untuk different detection methods
    detection_colors = {
        'contour': '#FF6B6B',     # Red
        'hough': '#4ECDC4',       # Teal
        'template': '#45B7D1',    # Blue
        'fusion': '#96CEB4',      # Green
        'ground_truth': '#FECA57' # Yellow
    }
    
    return detection_colors

colors = setup_visualization_style()

def visualize_detection_results(image, detections, title="Detection Results"):
    """Visualize detection results dari multiple methods"""
    fig, axes = plt.subplots(1, len(detections) + 1, figsize=(5 * (len(detections) + 1), 5))
    if len(detections) == 0:
        axes = [axes]
    
    # Original image
    axes[0].imshow(image, cmap='gray')
    axes[0].set_title('Original Image')
    axes[0].axis('off')
    
    # Detection results
    for idx, (method, detection_data) in enumerate(detections.items(), 1):
        if idx < len(axes):
            axes[idx].imshow(image, cmap='gray')
            
            # Overlay detection results berdasarkan method
            if 'grid_coordinates' in detection_data:
                coords = detection_data['grid_coordinates']
                if coords is not None:
                    # Draw bounding box
                    rect = plt.Rectangle(
                        (coords[0], coords[1]), 
                        coords[2] - coords[0], 
                        coords[3] - coords[1],
                        linewidth=2, 
                        edgecolor=colors.get(method, 'white'), 
                        facecolor='none'
                    )
                    axes[idx].add_patch(rect)
            
            confidence = detection_data.get('confidence', 0.0)
            axes[idx].set_title(f'{method.title()} (conf: {confidence:.3f})')
            axes[idx].axis('off')
    
    plt.suptitle(title, fontsize=16)
    plt.tight_layout()
    return fig

def log_detection_performance(method_name, processing_time, success, confidence=None):
    """Log detection performance metrics"""
    status = "SUCCESS" if success else "FAILED"
    conf_str = f", confidence: {confidence:.3f}" if confidence else ""
    
    logger.info(
        f"Detection [{method_name}] - Status: {status}, "
        f"Time: {processing_time:.3f}s{conf_str}"
    )

print("✅ Visualization utilities initialized")
print(f"✅ Detection colors: {list(colors.keys())}")
print("✅ Logging framework active")
```

## OpenCV Methods Validation

```python
# Validate critical OpenCV methods untuk template detection
def validate_opencv_methods():
    """Validate availability dan functionality OpenCV methods"""
    
    validation_results = {}
    
    # Test image untuk validation
    test_image = np.zeros((100, 100), dtype=np.uint8)
    cv2.rectangle(test_image, (20, 20), (80, 80), 255, 2)
    
    # 1. Contour detection
    try:
        contours, hierarchy = cv2.findContours(test_image, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
        validation_results['findContours'] = f"✅ Available - Found {len(contours)} contours"
    except Exception as e:
        validation_results['findContours'] = f"❌ Error: {str(e)}"
    
    # 2. Hough Transform
    try:
        edges = cv2.Canny(test_image, 50, 150)
        lines = cv2.HoughLines(edges, 1, np.pi/180, threshold=30)
        line_count = len(lines) if lines is not None else 0
        validation_results['HoughLines'] = f"✅ Available - Found {line_count} lines"
    except Exception as e:
        validation_results['HoughLines'] = f"❌ Error: {str(e)}"
    
    # 3. Template matching
    try:
        template = test_image[30:70, 30:70]
        result = cv2.matchTemplate(test_image, template, cv2.TM_CCOEFF_NORMED)
        min_val, max_val, min_loc, max_loc = cv2.minMaxLoc(result)
        validation_results['matchTemplate'] = f"✅ Available - Max correlation: {max_val:.3f}"
    except Exception as e:
        validation_results['matchTemplate'] = f"❌ Error: {str(e)}"
    
    # 4. Perspective transform
    try:
        src_points = np.float32([[0, 0], [100, 0], [100, 100], [0, 100]])
        dst_points = np.float32([[10, 10], [90, 10], [90, 90], [10, 90]])
        transform_matrix = cv2.getPerspectiveTransform(src_points, dst_points)
        validation_results['getPerspectiveTransform'] = "✅ Available - Transform matrix created"
    except Exception as e:
        validation_results['getPerspectiveTransform'] = f"❌ Error: {str(e)}"
    
    # 5. Morphological operations
    try:
        kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (5, 5))
        opened = cv2.morphologyEx(test_image, cv2.MORPH_OPEN, kernel)
        validation_results['morphologyEx'] = "✅ Available - Morphological operations working"
    except Exception as e:
        validation_results['morphologyEx'] = f"❌ Error: {str(e)}"
    
    return validation_results

opencv_validation = validate_opencv_methods()

print("🔍 OpenCV Methods Validation:")
for method, result in opencv_validation.items():
    print(f"  {method}: {result}")

# Count successful validations
success_count = sum(1 for result in opencv_validation.values() if result.startswith('✅'))
total_count = len(opencv_validation)

print(f"\n📊 Validation Summary: {success_count}/{total_count} methods validated successfully")

if success_count == total_count:
    print("✅ All OpenCV methods are ready untuk template detection implementation")
else:
    print("⚠️ Some OpenCV methods have issues - check environment setup")
```
